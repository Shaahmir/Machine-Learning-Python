# Reinforcement Learning - SARSA

## Importing the Libraries

```python
import random
import gymnasium as gym
import numpy as np
```

## Creating an Enviroment

```python
env = gym.make('CliffWalking-v1')
```

## Visualising Enviroment

```python
print(env.observation_space)
print(env.action_space)
```

```python
start, info = env.reset()
print(start, info)
```

## Defining Variables

```python
alpha = 0.5
gamma = 0.99
epsilon = 0.1
episodes = 500
```

## Initialising Q-Table

```python
Q = np.zeros((48, 4))
```

## ε-greedy policy - Exploration / Exploitation

```python
def e_greedy_policy(state):

    if random.random() < epsilon:
        return env.action_space.sample()

    q_values = Q[state]
    best_actions = np.flatnonzero(q_values == q_values.max())
    return random.choice(best_actions)
```

```python
def greedy_policy(state):
    
    q_values = Q[state]
    best_actions = np.flatnonzero(q_values == q_values.max())
    return random.choice(best_actions)
```

## Training the Agent

```python
for episode in range(episodes):

    state, _ = env.reset()
    action = e_greedy_policy(state)

    done = False
    total_reward = 0
    episode_len = 0

    while not done:

        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated

        if done:
            target = reward
        else:
            next_action = e_greedy_policy(next_state)
            target = reward + gamma * Q[next_state, next_action]

        Q[state, action] += alpha * (target - Q[state, action])

        if not done:
            state = next_state
            action = next_action

        total_reward += reward
        episode_len += 1
    
    if episode < 5 or episode == episodes - 1:

        print(f"Episode: {episode + 1}/{episodes}")
        print(f"Total Reward: {total_reward}")
        print(f"Episode Length: {episode_len}")
        print()

env.close()
```

## Visualising the Agent

```python
env = gym.make('CliffWalking-v1', render_mode = "human")
state ,_ = env.reset()

done = False
total_reward = 0
episode_len = 0

while not done:
    action = greedy_policy(state)
    next_state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated

    if not done:
        state = next_state

    total_reward += reward
    episode_len += 1

print(f"Total Reward: {total_reward}")
print(f"Episode Length: {episode_len}")

env.close()
```
